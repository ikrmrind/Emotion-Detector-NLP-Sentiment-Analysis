### Removing the Punchtuations

```python
import string
def remove_punch(txt):
    return txt.translate(str.maketrans('','',string.punctuation))
td['Text']=td['Text'].apply(remove_punch)    
```

### Removing Numbers 

```python
def remove_numbers(txt):
    new=''
    for i in txt:
        if not i.isdigit():
            new=new+i
    return new       
td['Text']=td['Text'].apply(remove_numbers)            
```

```python
td['Text']=td['Text'].apply(lambda x : x.lower())
td['Text']=td['Text'].str.replace(r'https', '',regex=True)
```

### Removing Emjoes

```python
def remove_emojies(txt):
    new=''
    for i in txt:
        if i.isascii():
            new= new + i
    return new

td['Text']=td['Text'].apply(remove_emojies)
```

### Removig Stop Words 

```python
#Ste of Stop Words prebuild (Stopword)
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
```

```python
#Downloading Stopwords
nltk.download('punkt')
nltk.download('stopwords') 
nltk.download('punkt_tab')
#many Stop words will be downloaded LIke Urdu English Frence Etc
```

```python
#We Just Need English Stop Eords
stop_words=set(stopwords.words('english')) #downloads the English Stop words (is , a the )
```

```python
len(stop_words)
```

```python
#now we need to make token of text TOkanize
#Match the Text by StoWords 
#Remeve Matching Words From Text
```

```python
def cleaned_text(txt):
    words=word_tokenize(txt) #words = txt.split()
    results=[]
    for i in words:
        if not i in stop_words:
            results.append(i)
    return ' '.join(results)
td['Text']=td['Text'].apply(cleaned_text)    
```

### Train Test Split

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(td['Text'],td['Numeric_Emotions'],test_size=0.2,random_state=42)
```

```python
### Using MutliNomal Nave Byes for Classifcation in NLP
```

### BAG OF WORD 

```python
from sklearn. feature_extraction. text import CountVectorizer, TfidfVectorizer

from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score

bow_vectorizer = CountVectorizer()
X_train_bow = bow_vectorizer.fit_transform(x_train)
X_test_bow = bow_vectorizer.transform(x_test)

nb_model = MultinomialNB()
nb_model.fit(X_train_bow, y_train)

pred_bow = nb_model.predict(X_test_bow)
print(accuracy_score(y_test, pred_bow))
```

### TF-IDF

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score,classification_report

tf_idf_vectorizer = TfidfVectorizer(ngram_range=(1,2))
X_train_TF_IDF = tf_idf_vectorizer.fit_transform(x_train)
X_test_TF_IDF = tf_idf_vectorizer.transform(x_test)

nb_model = MultinomialNB()
nb_model.fit(X_train_TF_IDF, y_train)

pred_TFidf = nb_model.predict(X_test_TF_IDF)
print(accuracy_score(y_test, pred_TFidf))
```

### TF-IDF with Logistic regression

```python
from sklearn.linear_model import LogisticRegression

tf_idf_vectorizer = TfidfVectorizer(lowercase=True)
X_train_TF_IDF = tf_idf_vectorizer.fit_transform(x_train)
X_test_TF_IDF = tf_idf_vectorizer.transform(x_test)

model = LogisticRegression().fit(X_train_TF_IDF, y_train)
pred_TFidf = model.predict(X_test_TF_IDF)
print(accuracy_score(y_test, pred_TFidf))
print(classification_report(y_test,pred_TFidf))
```
